# 1. KINEMATICS OF LIV

LIV can be phenomenologically described through a modified dispersion relation for photons, which is often expressed as:

$ E^2 = p^2 c^2 \times  \left[  1 + \sum_{n=1}^{\infty} S_n  \left( \frac{E}{E_{QG, n}} \right) ^n \right] $

$E_{QG,n}$ is the Quantum Gravity (QG) energy scale at which the LIV effects will be significant.

We will test linear ($n=1$) and quadratic ($n = 2$) effects in LIV studies.

$E$ and $p$ are respectively the energy and momentum of a photon. 

$c$ is the Lorentz invariant speed of light in vacuum

$S_n$ is a factor that could be $\pm 1$ depending if we are dealing with a subluminal scenario ($S_n=-1$), or a superluminal scenario ($S_n=+1$).

We can parameterize the LIV contribution by expressing $E_{QG,n}$ in terms of the Planck energy as: $ E_{QG, n} = \xi_n^{-1} \times E_{pl} $;

where $\xi_n$ will be named and reffered as the LIV parameter.

```python
##Import python libraries 
import numpy as np
import matplotlib.pyplot as plt
import astropy.units as u


## Import functions from liv framework
from liv_framework.liv import *
```

The kinematics of the $\gamma_{EBL} \; \gamma \rightarrow e^+ \; e^-$ interaction could be modified by LIV. 

A shift in the energy threshold of the interaction is given by:

\begin{equation}
    \epsilon_{th} = \frac{m_e^2 c^4}{E_{\gamma} (1 - \cos\theta)} - S_{n} \frac{E_{\gamma}^{n+1}}{4 E_{QG, n}^n }
\end{equation}

Where $\epsilon$ is the energy of the EBL photon and $E_{\gamma}$ the energy of the gamma-ray photon in the interaction.

Depending on the LIV scenario, we could have linear or quadratic effects in a subluminal or superluminal scenario.

```python
## Example: Plot for standard (n=0; No LIV) and linear (n=1) scenario for different values of the LIV parameter

plot_epsilon_threshold(n_values=[0, 1], xi_n_values=[1e8, 1e6, 1e4, 1e2, 1e0, 1e-2, 1e-4, 1e-6, 1e-8], liv_type="subluminal")
```

# 2. COMPUTATION OF LIV EFFECTS IN THE EBL

The modified energy threshold for $\epsilon$ has to be incorporated with some other changes into a triple integral to compute the EBL+LIV opacity:


\begin{equation}
    \begin{aligned}
        \tau_{\gamma\gamma}(E_\gamma, z, \xi_n) = \frac{c}{8E_{\gamma}^2} \int_0^z  \, \frac{dz'}{H(z') (1+z')^3} 
        \times \int_{\epsilon_{\text{min}}}^{\infty}  \, \frac{n(\epsilon, z')}{\epsilon^2} d\epsilon \int_{s_{\rm min} (z')}^{s_{\rm max} (z') }  \, [s- (m_{\gamma} c^2)^2 ] \, \sigma_{\gamma\gamma}(s) \, ds \; ,
    \end{aligned}
\end{equation}

$n(\epsilon, z')$ is the EBL photon number density

$\sigma_{\gamma\gamma}$ is the pair-production cross section as a function of $s = (m_{\gamma} c^2)^2  + 2 E_\gamma \epsilon (1 - cos\theta)$

An effective mass term for photons is defined as:
\begin{equation}
    (m_{\gamma} c^2)^2 = S_n \frac{ [ E_{\gamma} (z+1) ]^{n+2}}{E_{QG, n}^n}  \; ,
\end{equation}

and the limits of the integral in $s$ are defined as:
\begin{equation}
    s_{\rm min} = 4(m_e c^2)^2 \; ,
\end{equation}

and
\begin{equation}
    s_{\rm max} (z') = 4 \epsilon E_{\gamma} (1+z') + (m_{\gamma} c^2)^2 \; .
\end{equation}


```python
## Import functions from liv framework
from liv_framework.ebl import *
```

For the 1st integral, we will need a function $\sigma_{\gamma\gamma}(s)$ to compute the cross-section of the interaction as a function of s:

```python
# Define constants
E_gamma = 10 * u.TeV  # Gamma-ray energy
epsilon = 1 * u.eV   # Target photon energy
z = 0.997             # Redshift
xi_n = 0.01          # LIV parameter
sigma_T = 6.6524e-25  # Thomson cross-section in cm^2
prefactor = (3/16) * sigma_T  # cm^2

# Compute E_LIV
E_pl = (1.22e19 * u.GeV).to(u.eV)  # Planck energy in eV
E_LIV = xi_n * E_pl  # LIV energy scale

# Compute s_min and s_max
s_min = (4 * const.m_e**2 * const.c**4).to(u.eV**2)  # Threshold for pair production
s_max = (4 * E_gamma * epsilon * (1 + z) - (E_gamma**3 * (1 + z)**3 / E_LIV)).to(u.eV**2) #s_max for subluminal scenario

# Generate s values for plotting
s_values = np.logspace(np.log10(s_min.value), np.log10(s_max.value), 100)
sigma_values = [prefactor * sigma_breit_wheeler_s_numba(s) for s in s_values]

# Plot the Breit-Wheeler cross-section
plt.figure(figsize=(8, 6))
plt.loglog(s_values, sigma_values, linewidth=2, color='blue', label=r"$\sigma_{\gamma\gamma}(s)$")
plt.xlabel(r"$s$ [eV²]")
plt.ylabel(r"$\sigma_{\gamma\gamma}$ [cm²]")
plt.title("Breit-Wheeler Cross Section")
plt.legend()
plt.grid(True, which="both", linestyle="--", alpha=0.6)
plt.show()
```

For the third integral in z, there is a cosmological expansion factor, here we assume $H(z')$ is the Hubble parameter for standard  $\Lambda$CDM cosmology $H(z') = H_0 \sqrt{\Omega_m (1+z')^3 + \Omega_\Lambda}$ with $H_0 = 70$ km s$^{-1}$ Mpc$^{-1}$, $\Omega_m = 0.3$, $\Omega_\Lambda = 0.7$.


Now, we can compute the triple integral with the LIV modifications to obtain the EBL+LIV model for Dominguez et al. 2011

```python
# ==============================================================
# LOAD PHOTON ENERGY DENSITY FROM DOMINGUEZ EBL MODEL
# ==============================================================

# --- Dominguez et al. 2011 model ---
n_interp_dom, epsilon_grid_dom = get_ebl_interpolator(
    model="dominguez",
    file_path="../data/ebl_intensity_dominguez11.txt")
```

```python
## Load for computation in paralel
from concurrent.futures import ProcessPoolExecutor

# ---------------------------------------------------------
# MAIN FUNCTIONS FOR COMPUTATION OF EBL OPACITY WITH LIV EFFECTS 
# (THESE FUNCTIONS COULD BE IMPORTED FROM 'liv_framework.ebl')
# ---------------------------------------------------------
def tau_liv(E_gamma, z, n_interp, epsilon_grid, xi_n=0.0, n_order=1):
    """
    Optical depth including LIV effects using comoving EBL density.

    xi_n:
        = 0  → standard EBL
        > 0  → superluminal
        < 0  → subluminal
    """

    # --- Energy ---
    E_gamma = E_gamma.to(u.eV)
    E_gamma_val = E_gamma.value

    # --- Constants ---
    s_thresh = (4 * const.m_e**2 * const.c**4).to(u.eV**2).value
    sigma_T = 6.6524e-25

    prefactor = (3/16) * sigma_T

    # --- LIV handling ---
    if xi_n == 0:
        sign = 0
    else:
        sign = np.sign(xi_n)
        E_pl = (1.22e19 * u.GeV).to(u.eV)
        E_LIV = (E_pl / abs(xi_n)).to(u.eV).value
        

    def delta_liv(E_local):
        if xi_n == 0:
            return 0.0
        if n_order == 1:
            return sign * (E_local**3) / E_LIV
        elif n_order == 2:
            return sign * (E_local**4) / (E_LIV**2)
        else:
            raise ValueError("n_order must be 1 or 2")
            
    # ---------------------------------------------------------
    # Vacuum photon decay for superluminal LIV
    # ---------------------------------------------------------

    if sign > 0:

        E_max_path = E_gamma_val * (1 + z)

        if delta_liv(E_max_path) >= s_thresh:
            return 1e+30  #huge value instead of +inf 


    # --- Integrals ---
    def redshift_integral(z_int):
        E_local = E_gamma_val * (1 + z_int)

        def epsilon_integral_log(log_epsilon):
            epsilon = np.exp(log_epsilon)
            
            # Simplified EBL lookup: comoving 
            try:
                n_epsilon = n_interp((np.log(epsilon), z_int))
            except:
                return 0.0
            
            
            delta = delta_liv(E_local)

            s_min = s_thresh
            s_max = 4 * E_local * epsilon + delta

            if s_max <= s_min:
                return 0.0
            
            ### condition if there is a negligible contribution
            if np.log(s_max) - np.log(s_min) < 1e-8:
                return 0.0

            
            def s_integral_log(u):
                s = np.exp(u)
                sigma = sigma_breit_wheeler_s_numba(s)
                return sigma * (s - delta) * s  # extra s from log-substitution

            u_min = np.log(s_min)
            u_max = np.log(s_max)

            s_int, _ = quad(s_integral_log, u_min, u_max,
                            epsabs=1e-4, epsrel=1e-3, limit=60)

            return (n_epsilon / epsilon) * s_int  # 1/epsilon from log-substitution

        delta = delta_liv(E_local)
        epsilon_min = (s_thresh - delta) / (4 * E_local)
        epsilon_max = np.max(epsilon_grid)

        if epsilon_min >= epsilon_max:
            return 0.0

        if epsilon_min <= 0:
            epsilon_min = max(epsilon_min, epsilon_grid[0])   # this is a small positive value

        epsilon_int, _ = quad(
            epsilon_integral_log,
            np.log(epsilon_min),
            np.log(epsilon_max),
            epsabs=1e-4, epsrel=1e-3, limit=60
        )

        # Cosmology
        Omega_m, Omega_lambda = 0.3, 0.7
        H_z = np.sqrt(Omega_m * (1 + z_int)**3 + Omega_lambda)

        return epsilon_int / H_z
    
    # --- Outer integral ---
    int_final, _ = quad(redshift_integral, 0, z,
                        epsabs=1e-4, epsrel=1e-3, limit=60)

    H0 = 70 * (u.km / u.s / u.Mpc)

    final_prefactor = (
        (prefactor / (8 * E_gamma_val**2)) *
        (const.c / H0).to(u.cm) *
        (u.eV**2 / u.cm)
    ).value

    return final_prefactor * int_final
    


# ===============================
# MULTIPROCESSING FUNCTION
# ===============================
def compute_tau_single(args):
    """Unpack arguments and call simplified tau_liv."""
    E, xi_val, z_val, n_int, eps_grid, order = args
    return tau_liv(E, z_val, n_int, eps_grid, xi_n=xi_val, n_order=order)


# ===============================
# PARALLEL COMPUTATION
# ===============================
def compute_model(n_interp, epsilon_grid, model_name, xi_n_array, E_gamma_vals, z, n_order=1):

    model_results = []

    # Parallel processing
    with ProcessPoolExecutor() as executor:

        for xi_n_val in xi_n_array:

            print(f"[{model_name}] Processing xi_n = {xi_n_val:.3e}")

            args_list = [
                (
                    E,
                    xi_n_val,
                    z,
                    n_interp,
                    epsilon_grid,
                    n_order,
                )
                for E in E_gamma_vals
            ]

            tau_vals = list(
                executor.map(
                    compute_tau_single,
                    args_list
                )
            )

            model_results.append(tau_vals)

    return np.asarray(model_results)
```

```python
# ===============================
# SET PARAMETERS FOR COMPUTATION
# ===============================

## Energy range in TeVs
E_gamma_vals = np.logspace(-1.8, 2, 40) * u.TeV
E_gamma_TeV = E_gamma_vals.to_value(u.TeV)

## LIV paramter values to compute: could be standard (=0), subluminal (-) and/or superluminal (+)
xi_n_array = [-1e5, -1e3, 0, 1e3, 1e5] 

## redshift 
z = 1.0

## Linear (n_order=1) or Quadratic (n_order=2) effects
n_order = 1
```

```python
# ===============================
# COMPUTE EBL+LIV Opacity
# ===============================

## This array will contain the computed EBL+LIV opacity for the Dominguez model 
results_dominguez = compute_model(n_interp_dom, epsilon_grid_dom, "Dominguez", xi_n_array, E_gamma_vals, z, n_order=1)
```

```python
from matplotlib.lines import Line2D
from scipy.interpolate import PchipInterpolator

# ===============================
# PLOTTING
# ===============================

# ==========================================
# FIGURE
# ==========================================
fig, ax = plt.subplots(figsize=(9, 7))

# ==========================================
# COLOR MAPS
# ==========================================
sub_vals = [x for x in xi_n_array if x < 0]
sup_vals = [x for x in xi_n_array if x > 0]
    
colors_sub = plt.cm.Blues(np.linspace(0.5, 0.9, len(sub_vals)))
colors_sup = plt.cm.Reds(np.linspace(0.5, 0.9, len(sup_vals)))

# ==========================================
# PLOT: DOMINGUEZ MODEL EBL+LIV
# ==========================================
sub_idx = 0
sup_idx = 0

for i, xi_n_val in enumerate(xi_n_array):

    interp_tau = PchipInterpolator(
        E_gamma_vals.value,
        results_dominguez[i],
        extrapolate=False
    )

    if xi_n_val == 0:

        ax.plot(
            E_gamma_TeV,
            results_dominguez[i],
            color='black',
            lw=3,
            label="Standard (No LIV)"
        )

    elif xi_n_val < 0:

        ax.loglog(
            energies_interp,
            interp_tau(energies_interp),
            color=colors_sub[sub_idx],
            lw=2.5,
            ls='--',
            label=f"$\\xi_1$ = {xi_n_val:.1e}"
        )

        sub_idx += 1

    else:

        ax.loglog(
            energies_interp,
            interp_tau(energies_interp),
            color=colors_sup[sup_idx],
            lw=2.5,
            ls='--',
            label=f"$\\xi_1$ = {xi_n_val:.1e}"
        )

        sup_idx += 1
# ==========================================
# AXES STYLE
# ==========================================
ax.set_xscale("log")
ax.set_yscale("log")

ax.set_xlim(E_gamma_TeV.min(), E_gamma_TeV.max())
ax.set_ylim(1e-4, 1e4)

ax.set_xlabel(r"$E_{\gamma}$ [TeV]")
ax.set_ylabel(r"$\tau_{\mathrm{LIV}} (E_{\gamma}, z=1.0)$")

# ==========================================
# GRID
# ==========================================
ax.grid(True, which="major", linestyle="--", alpha=0.6)
ax.grid(True, which="minor", linestyle=":", alpha=0.3)

# ==========================================
# LEGEND
# ==========================================
legend = ax.legend(
    fontsize=12,
    loc=3,
    frameon=True,
    fancybox=True,
    framealpha=0.95,
    borderpad=0.8
)

# ==========================================
# OPTIONAL TITLE
# ==========================================
#ax.set_title("Dominguez et al. 2011 Model Opacity (z={z}, n={n_order})', fontsize=18)

# ==========================================
# TICKS
# ==========================================
ax.tick_params(which='both', width=1.5)
ax.tick_params(which='major', length=10)
ax.tick_params(which='minor', length=4)

# ==========================================
# LAYOUT
# ==========================================
plt.tight_layout()

# ==========================================
# SAVE FIGURE
# ==========================================
#plt.savefig(
#    "Dominguez_EBL_opacity_linear_LIV_example.png",
#    dpi=300,
#    bbox_inches="tight"
#)

plt.show()
```
